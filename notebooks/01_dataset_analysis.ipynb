# Hyperspectral Object Detection Challenge 2026
## Notebook 01: Dataset Analysis & Inspection
This notebook analyzes the 3,000 training scenes, VOC XML annotations, class balance across all 18 categories, and object scale distributions.

```python
import os
import sys
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Ensure project root in path
sys.path.insert(0, "..")
from datasets import CLASS_NAMES, PAIRED_CATEGORIES, parse_voc_annotation

print(f"Loaded {len(CLASS_NAMES)} target categories:")
for i, c in enumerate(CLASS_NAMES):
    print(f"  [{i:2d}] {c}")
```

```python
# Inspect VOC XML annotation distribution
ann_files = glob.glob("../data/raw/annotations/*.xml")
if not ann_files:
    ann_files = glob.glob("../data/raw/train/*.xml")

print(f"Found {len(ann_files)} VOC XML annotations.")

class_counts = {c: 0 for c in range(len(CLASS_NAMES))}
box_areas = []
aspect_ratios = []

for fp in ann_files:
    data = parse_voc_annotation(fp)
    for box, label in zip(data["boxes"], data["labels"]):
        class_counts[label] += 1
        w = box[2] - box[0]
        h = box[3] - box[1]
        box_areas.append(w * h)
        aspect_ratios.append(w / max(1.0, h))

# Plot Class Distribution
plt.figure(figsize=(12, 5))
plt.bar(range(18), [class_counts[i] for i in range(18)], color="royalblue")
plt.xticks(range(18), CLASS_NAMES, rotation=90)
plt.ylabel("Instance Count")
plt.title("Class Frequency Distribution across 18 Categories")
plt.tight_layout()
plt.show()
```
