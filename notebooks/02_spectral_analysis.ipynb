# Hyperspectral Object Detection Challenge 2026
## Notebook 02: Spectral Signature Analysis & Pseudo-RGB Visualization
Investigates the 16 continuous spectral bands between 460 nm and 600 nm and compares spectral curves between authentic and counterfeit material pairs.

```python
import os
import sys
import numpy as np
import matplotlib.pyplot as plt

sys.path.insert(0, "..")
from datasets import WAVELENGTHS, CLASS_NAMES, PAIRED_CATEGORIES, render_pseudo_rgb, render_false_color, plot_spectral_signature

print(f"16 Spectral Wavelengths (nm): {WAVELENGTHS}")
```

```python
# Load a sample 16-band cube
from datasets.hyperspectral_dataset import HyperspectralDataset

ds = HyperspectralDataset("../data/raw", synthetic_count=2)
cube_tensor, target = ds[0]
cube_hwc = np.transpose(cube_tensor.numpy(), (1, 2, 0))

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
pseudo_rgb = render_pseudo_rgb(cube_hwc)
false_color = render_false_color(cube_hwc, r_band=15, g_band=8, b_band=2)

axes[0].imshow(pseudo_rgb)
axes[0].set_title("Natural Pseudo-RGB Composite (R:15, G:7, B:1)")
axes[0].axis("off")

axes[1].imshow(false_color)
axes[1].set_title("False-Color Composite (R:15, G:8, B:2)")
axes[1].axis("off")
plt.tight_layout()
plt.show()
```

```python
# Plot spectral reflectance curves over detected object bounding boxes
fig = plot_spectral_signature(
    cube_hwc,
    target["boxes"].numpy().tolist(),
    target["labels"].numpy().tolist(),
    CLASS_NAMES,
    wavelengths=WAVELENGTHS
)
plt.show()
```
